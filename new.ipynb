```python
def get_chatbot_response(client, model_name, messages):
    # Add Markdown instruction to the messages to ensure output is in Markdown format
    markdown_instruction = " Please format the answer in Markdown with each description as regular text."
    
    # Update the last message (user's message) to include the markdown instruction
    messages[-1]["content"] += markdown_instruction

    # Prepare the input messages with roles and content
    input_messages = []
    for message in messages:
        input_messages.append({"role": message["role"], "content": message["content"]})

    # Generate the response using the provided client and model
    response = client.chat.completions.create(
        model=model_name,
        messages=input_messages,
        
        top_p=0.8,
        max_tokens=2000,
    ).choices[0].message.content
    
    return response
```

```python
messages = [
    {"role": "user", "content": "What's the difference between a coffee machine and a coffee shop?"}
]

response = get_chatbot_response(client, model_name, messages)
print(response)
```
