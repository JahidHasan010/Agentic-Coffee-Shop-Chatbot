```python
# Define the function to get chatbot response
def get_chatbot_response(client, model, messages, temperature=0):
    # Since 'messages' is a list of dictionaries with roles and content,
    # convert it to a simple prompt string for the model
    prompt = " ".join([message["content"] for message in messages])  # Joining all message contents as a single prompt
    
    # Generate the response using the model
    response = model.generate_content(
        prompt=prompt,  # The prompt is a string (not a list of messages)
        temperature=temperature,
        max_tokens=2000
    )
    
    # Return the generated response text
    return response.text  # Assuming 'response.text' holds the generated text
```

```python
import os
import google.generativeai as genai

# Ensure the API key and model name environment variables are set correctly
google_api_key = os.getenv("GOOGLE_API_KEY")
model_name_str = os.getenv("MODEL_NAME")

# Ensure both variables are retrieved properly from environment
if not google_api_key or not model_name_str:
    raise ValueError("API Key or Model Name not set in environment variables")

# Configure the client with the API key
client = genai.configure(api_key=google_api_key)

# Check if client is properly initialized
if client is None:
    raise ValueError("Failed to initialize the client")

# Initialize the model with the desired name from environment variable
model_name = genai.GenerativeModel(model_name_str)

# Define the function to get chatbot response
def get_chatbot_response(client, model_name, messages, temperature=0):
    input_messages = []
    for message in messages:
        input_messages.append({"role": message["role"], "content": message["content"]})

    # Generate the response using the model
    response = model_name.generate_content(
        prompt=input_messages,  # Use 'prompt' or whatever key is required by the API for the message content
        temperature=temperature,
        max_tokens=2000
    )

    # Return the generated response
    return response.text  # Assuming 'response.text' holds the generated text

# Example messages to test the function
messages = [{'role': 'user', 'content': "What's the capital of Italy?"}]
response = get_chatbot_response(client, model_name, messages)
print(response)
```
